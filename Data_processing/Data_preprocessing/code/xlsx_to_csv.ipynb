```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
def redistribute_time(t):
    t = t.to_numpy().astype(float)
    n = len(t)

    i = 0
    while i < n:
        j = i + 1
        while j < n and t[j] == t[i]:
            j += 1

        block_len = j - i

        # если это реальный "пучок" и есть точки до и после
        if block_len > 1 and i > 0 and j < n:
            t_prev = t[i - 1]
            t_next = t[j]

            step = (t_next - t_prev) / (block_len + 1)

            for k in range(block_len):
                t[i + k] = t_prev + step * (k + 1)

        i = j

    return t
```

```python
current_dir = Path.cwd()
print(f"Current dir: {current_dir}")

project_root = current_dir.parent 
raw_data_dir = project_root / "raw_data"

for temp in ["+25", "+30", "+35"]:
    for i in range(1, 6):
        excel_file = raw_data_dir / temp / str(i) / f"105-1-{i}-{i} {temp}.xlsx"
        
        df = pd.read_excel(excel_file, sheet_name='record')

        df['t,s'] = pd.to_timedelta(df['Total Time']).dt.total_seconds()
        df['t,s'] = redistribute_time(df['t,s'])

        df['Step'] = (df['Step Type'] != df['Step Type'].shift()).cumsum()

        csv_file = excel_file.parent / f"{i} {temp}.csv"
        df.to_csv(
            csv_file,
            columns=['Step', 'Step Type', 't,s', 'Current(A)', 'Voltage(V)'],
            header=['Step', 'Step Type', 't,s', 'I,A', 'U,V'],
            index=False
        )

        print(f"{csv_file.relative_to(project_root)} was saved")
```
