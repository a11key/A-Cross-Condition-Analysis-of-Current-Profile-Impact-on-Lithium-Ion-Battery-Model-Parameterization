```python
import pandas as pd

for i in range (2, 6):
    # Путь к исходному файлу Excel
    input_file = f'D://Documents//battery//MJI//data//+25//{i}//105-1-{i}-{i}.xlsx'
    output_file = f'D://Documents//battery//MJI//data//+25//{i}//105-1-{i}-{i}.csv'
    
    # Чтение листа "record"
    df = pd.read_excel(input_file, sheet_name='record')
    
    # Преобразование времени в секунды
    df['t,s'] = pd.to_timedelta(df['Total Time']).dt.total_seconds()
    
    # Создание номера шага, увеличиваемого при смене Step Type
    step_counter = 0
    steps = []
    previous_step_type = None
    
    for st in df['Step Type']:
        if st != previous_step_type:
            step_counter += 1
            previous_step_type = st
        steps.append(f"{step_counter} {st}")
    
    df['Step'] = steps
    
    # Переименование остальных столбцов
    df = df.rename(columns={
        'Current(A)': 'I,A',
        'Voltage(V)': 'U,V'
    })
    
    # Сохранение в CSV
    df[['Step', 't,s', 'I,A', 'U,V']].to_csv(output_file, index=False)
    
    print(f"Данные успешно сохранены в {output_file}")
```
